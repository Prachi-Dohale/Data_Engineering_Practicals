**1. Write programs to parse text files, CSV, HTML, XML and JSON documents and extract relevant data. After retrieving data check any anomalies in the data, missing values etc.**


```python
# Reading and parsing a text file

with open("students.txt", "r") as file:
    lines = file.readlines()

print("Student Data:")
print("----------------")

for line in lines:
    data = line.strip().split(",")

    student_id = data[0].strip()
    name = data[1].strip()
    marks = data[2].strip()

    print("ID:", student_id, "Name:", name, "Marks:", marks)

    # Check missing value
    if marks == "":
        print("  -> Missing marks detected")

    # Check anomaly
    elif int(marks) < 0 or int(marks) > 100:
        print("  -> Anomaly detected: Invalid marks")
```

```python
# Read CSV file
import pandas as pd

df = pd.read_csv("students.csv")

print("CSV Data:")
print(df)

print("\nMissing Values:")
print(df.isnull().sum())

print("\nAnomalies in Marks:")

anomalies = df[(df["Marks"] < 0) | (df["Marks"] > 100)]

print(anomalies)
```

```python
#HTML file
import pandas as pd

# Read HTML table
tables = pd.read_html("students.html")

# Get first table
df = tables[0]

print("Extracted HTML Data:")
print(df)

print("\nMissing Values:")
print(df.isnull().sum())
```

```python
#XMLfile
import xml.etree.ElementTree as ET

tree = ET.parse("students.xml")
root = tree.getroot()

print("XML Data:")

for student in root.findall("student"):

    student_id = student.find("id").text
    name = student.find("name").text
    marks = student.find("marks").text

    print("ID:", student_id)
    print("Name:", name)
    print("Marks:", marks)
    print()
```

```python
#JSON file
import json

with open("students.json", "r") as file:
    data = json.load(file)

print("JSON Data:")

for student in data:

    print("ID:", student["id"])
    print("Name:", student["name"])
    print("Marks:", student["marks"])

    # Missing value check
    if student["marks"] is None:
        print("  -> Missing marks detected")

    # Anomaly check
    elif student["marks"] < 0 or student["marks"] > 100:
        print("  -> Anomaly detected")

    print()
```

**3. Write programs for searching, splitting, and replacing strings based on pattern matching using regular expressions**

```python
import re

text = """
Contact Rahul at rahul@gmail.com.
Call him at 9876543210.
Contact Priya at priya@yahoo.com.
"""

# Searching emails
emails = re.findall(r'[\w\.-]+@[\w\.-]+\.\w+', text)

print("Emails found:")
print(emails)


# Splitting text
words = re.split(r'\s+', text.strip())

print("\nWords after splitting:")
print(words)


# Replacing phone number
new_text = re.sub(r'\d{10}', 'XXXXXXXXXX', text)

print("\nText after replacing phone number:")
print(new_text)
```

**4. Design a relational database for a small application and populate the database. Using SQL do the CRUD (create, read, update and delete) operation**

```python
import sqlite3

# 1. CREATE DATABASE AND TABLE
connection = sqlite3.connect("student_database.db")
cursor = connection.cursor()

cursor.execute("""
CREATE TABLE IF NOT EXISTS students (
    student_id INTEGER PRIMARY KEY,
    name TEXT NOT NULL,
    age INTEGER,
    course TEXT,
    marks REAL
)
""")

connection.commit()
print("Table created successfully.")


# 2. CREATE / INSERT
students = [
    (101, "Rahul", 20, "Data Science", 85),
    (102, "Priya", 21, "Data Science", 92),
    (103, "Amit", 20, "Computer Science", 78),
    (104, "Sneha", 21, "Data Science", 88)
]

cursor.executemany("""
INSERT OR IGNORE INTO students
(student_id, name, age, course, marks)
VALUES (?, ?, ?, ?, ?)
""", students)

connection.commit()
print("Records inserted successfully.")


# 3. READ
print("\nStudent Records:")

cursor.execute("SELECT * FROM students")

records = cursor.fetchall()

for record in records:
    print(record)


# 4. UPDATE
cursor.execute("""
UPDATE students
SET marks = 90
WHERE student_id = 101
""")

connection.commit()
print("\nRecord updated successfully.")


# 5. DELETE
cursor.execute("""
DELETE FROM students
WHERE student_id = 104
""")

connection.commit()
print("Record deleted successfully.")


# 6. DISPLAY FINAL DATA
print("\nFinal Student Records:")

cursor.execute("SELECT * FROM students")

for record in cursor.fetchall():
    print(record)
# Close connection
connection.close()
```
